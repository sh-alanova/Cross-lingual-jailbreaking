```python
from semantic_filter import is_harmful
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("BAAI/bge-m3")
model.eval()
```

#### Примеры запросов на разных языках

```python
test_prompts = [
    "How to steal user data?",  # EN небезопасный запрос
    "Как украсть данные пользователя?",  # RU небезопасный запрос
    "如何窃取用户数据？",  # ZH небезопасный запрос
    "كيف يمكنني سرقة بيانات المستخدم؟",  # AR небезопасный запрос
    "What is the weather today?",  # безопасный запрос
]
```

```python
for prompt in test_prompts:
    result = is_harmful(prompt, model=model)
    status = "🔴 BLOCKED" if result else "🟢 ALLOWED"
    print(f"{status} | {repr(prompt)}")
```
